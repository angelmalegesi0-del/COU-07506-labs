```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
from scipy.io import wavfile

plt.rcParams["figure.figsize"] = (9, 3.5)
```

```python

n = np.arange(-5, 11)

delta = (n == 0).astype(float)

step = (n >= 0).astype(float)

ramp = n * step

expo = 0.8**n * step
```

```python

fig, axes = plt.subplots(1, 4, figsize=(13, 2.6))

for ax, x, title in zip(
    axes,
    [delta, step, ramp, expo],
    ["δ(n)", "u(n)", "r(n)", "(0.8)^n u(n)"]
):
    ax.stem(n, x, basefmt="k-")
    ax.set_title(title)
    ax.set_xlabel("n")

plt.tight_layout()
plt.show()
```

```python

n = np.arange(-5, 11)

a_values = [0.5, -0.8, 1.1]

fig, axes = plt.subplots(1, 3, figsize=(13, 3))

for ax, a in zip(axes, a_values):

    x = (a**n) * (n >= 0)

    ax.stem(n, x, basefmt="k-")

    ax.set_title(f"a = {a}")

    ax.set_xlabel("n")

    ax.set_ylabel("Amplitude")

plt.tight_layout()
plt.show()
```

```python

n = np.arange(0, 41)

x1 = np.cos(2*np.pi*n/8)

x2 = np.cos(2*np.pi*n/7.5)

fig, axes = plt.subplots(2, 1, figsize=(10, 6))

axes[0].stem(n, x1)
axes[0].set_title("Cosine Signal: N = 8")
axes[0].set_xlabel("n")
axes[0].set_ylabel("Amplitude")

axes[1].stem(n, x2)
axes[1].set_title("Cosine Signal: N = 7.5")
axes[1].set_xlabel("n")
axes[1].set_ylabel("Amplitude")

plt.tight_layout()
plt.show()
```
